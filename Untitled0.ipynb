```python
from google.colab import drive
drive.mount('/content/drive')

import os
import gc
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

CHECKPOINT_DIR = "/content/drive/MyDrive/loan_project_checkpoints"
os.makedirs(CHECKPOINT_DIR, exist_ok=True)

def save_checkpoint(name, **objects):
    path = os.path.join(CHECKPOINT_DIR, f"{name}.joblib")
    joblib.dump(objects, path)
    print("Saved:", path)

def load_checkpoint(name):
    path = os.path.join(CHECKPOINT_DIR, f"{name}.joblib")
    if os.path.exists(path):
        print("Loading:", path)
        return joblib.load(path)
    print("No checkpoint at:", path)
    return None
```

```python
RAW_PATH = "/content/drive/MyDrive/Lending_Club_Credit_Risk/00_raw_data/loan.csv"  # <-- update this

print("File exists:", os.path.exists(RAW_PATH))
print("Size (MB):", os.path.getsize(RAW_PATH) / 1e6)

df = pd.read_csv(RAW_PATH, low_memory=False)
print("Dataset shape:", df.shape)
```

```python
print("Loan status categories:")
print(df["loan_status"].value_counts(dropna=False))

print("\nDuplicate rows:", df.duplicated().sum())
print("Duplicate IDs:", df["id"].duplicated().sum())

print("\nMissing values — top 20:")
print(df.isna().sum().sort_values(ascending=False).head(20))
```

```python
df["issue_date"] = pd.to_datetime(df["issue_d"], format="%b-%Y", errors="coerce")
df["issue_year"] = df["issue_date"].dt.year

print("Issue year distribution:")
print(df["issue_year"].value_counts().sort_index())

df_cohort = df[df["issue_year"].between(2007, 2014)].copy()
del df
gc.collect()

print("\n2007-2014 cohort shape:", df_cohort.shape)
```

```python
default_statuses = [
    "Charged Off",
    "Does not meet the credit policy. Status:Charged Off"
]
non_default_statuses = [
    "Fully Paid",
    "Does not meet the credit policy. Status:Fully Paid"
]

df_model = df_cohort[df_cohort["loan_status"].isin(default_statuses + non_default_statuses)].copy()
del df_cohort
gc.collect()

df_model["default"] = df_model["loan_status"].isin(default_statuses).astype(int)

print("Modeling cohort shape:", df_model.shape)
print("Default rate: {:.2f}%".format(df_model["default"].mean() * 100))
print(df_model["default"].value_counts())
```

```python
leakage_cols = [
    "out_prncp", "out_prncp_inv", "total_pymnt", "total_pymnt_inv",
    "total_rec_prncp", "total_rec_int", "total_rec_late_fee",
    "recoveries", "collection_recovery_fee",
    "last_pymnt_d", "last_pymnt_amnt", "next_pymnt_d", "last_credit_pull_d",
    "hardship_flag", "hardship_type", "hardship_reason", "hardship_status",
    "deferral_term", "hardship_amount", "hardship_start_date", "hardship_end_date",
    "payment_plan_start_date", "hardship_length", "hardship_dpd", "hardship_loan_status",
    "orig_projected_additional_accrued_interest", "hardship_payoff_balance_amount",
    "hardship_last_payment_amount", "debt_settlement_flag", "debt_settlement_flag_date",
    "settlement_status", "settlement_date", "settlement_amount",
    "settlement_percentage", "settlement_term"
]
target_cols = ["loan_status", "default"]
identifier_cols = ["id", "member_id", "url", "desc", "zip_code", "title", "emp_title"]

constant_cols = [c for c in df_model.columns if df_model[c].nunique(dropna=False) <= 1]
all_missing_cols = [c for c in df_model.columns if df_model[c].isna().all()]

remove_cols = sorted(set(leakage_cols + target_cols + identifier_cols + constant_cols + all_missing_cols))

X_candidate = df_model.drop(columns=remove_cols, errors="ignore")
y = df_model["default"].copy()

print("Columns removed:", len(remove_cols))
print("Candidate predictor shape:", X_candidate.shape)
```

```python
missing_summary = pd.DataFrame({
    "missing_count": X_candidate.isna().sum(),
    "missing_pct": (X_candidate.isna().mean() * 100).round(2),
    "n_unique": X_candidate.nunique(dropna=True),
    "dtype": X_candidate.dtypes.astype(str)
}).sort_values("missing_pct", ascending=False)

display(missing_summary)
```

```python
audit = pd.DataFrame({
    "column": X_candidate.columns,
    "missing_pct": (X_candidate.isna().mean().values * 100).round(2),
})

def classify_missingness(pct):
    if pct == 0:
        return "no_missing"
    elif pct < 5:
        return "light"
    elif pct < 20:
        return "moderate"
    else:
        return "high"

audit["bucket"] = audit["missing_pct"].apply(classify_missingness)

high_missing_cols = audit.loc[audit["bucket"] == "high", "column"].tolist()
moderate_missing_cols = audit.loc[audit["bucket"] == "moderate", "column"].tolist()
light_missing_cols = audit.loc[audit["bucket"] == "light", "column"].tolist()

print("High:", len(high_missing_cols), "| Moderate:", len(moderate_missing_cols),
      "| Light:", len(light_missing_cols))
```

```python
X_flagged = X_candidate.copy()
flag_cols = high_missing_cols + moderate_missing_cols + light_missing_cols

for col in flag_cols:
    X_flagged[col + "_was_missing"] = X_flagged[col].isna().astype(int)

print("X_flagged shape:", X_flagged.shape)
print("Flags added:", len(flag_cols))
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_flagged, y, test_size=0.20, stratify=y, random_state=42
)
X_train = X_train.reset_index(drop=True)
X_test = X_test.reset_index(drop=True)
y_train = y_train.reset_index(drop=True)
y_test = y_test.reset_index(drop=True)

del X_candidate, X_flagged
gc.collect()

print("X_train:", X_train.shape, "| X_test:", X_test.shape)
print("Train default rate: {:.2f}%".format(y_train.mean() * 100))
print("Test default rate: {:.2f}%".format(y_test.mean() * 100))

save_checkpoint("step_split", X_train=X_train, X_test=X_test, y_train=y_train, y_test=y_test,
                high_missing_cols=high_missing_cols, moderate_missing_cols=moderate_missing_cols,
                light_missing_cols=light_missing_cols, flag_cols=flag_cols)
```

```python
train_eda = X_train.copy()
train_eda["default"] = y_train.values

key_cats = ["grade", "sub_grade", "purpose", "home_ownership", "verification_status", "term", "emp_length"]
for col in key_cats:
    print(f"\n=== Default rate by {col} ===")
    print(train_eda.groupby(col)["default"].agg(["mean", "count"])
          .rename(columns={"mean": "default_rate"}).sort_values("default_rate", ascending=False))

plt.figure(figsize=(8, 5))
sns.barplot(data=train_eda, x="grade", y="default", order=sorted(train_eda["grade"].unique()), errorbar=None)
plt.title("Default rate by loan grade (train set)")
plt.ylabel("Default rate")
plt.show()

del train_eda
gc.collect()
```

```python
from scipy.stats import chi2_contingency, mannwhitneyu

categorical_vars = ["grade", "sub_grade", "purpose", "home_ownership",
                     "verification_status", "term", "emp_length", "initial_list_status"]

chi_results = []
for col in categorical_vars:
    contingency = pd.crosstab(X_train[col], y_train)
    chi2, p_value, dof, expected = chi2_contingency(contingency)
    chi_results.append({"variable": col, "chi2_statistic": chi2, "p_value": p_value, "dof": dof})

chi_df = pd.DataFrame(chi_results).sort_values("chi2_statistic", ascending=False)
bonferroni_alpha = 0.05 / len(categorical_vars)
chi_df["significant_bonferroni"] = chi_df["p_value"] < bonferroni_alpha
print(f"Bonferroni-corrected alpha: {bonferroni_alpha:.6f}")
display(chi_df)

numeric_vars = ["loan_amnt", "int_rate", "dti", "annual_inc", "revol_util", "open_acc"]
mw_results = []
for col in numeric_vars:
    group0 = X_train.loc[y_train == 0, col].dropna()
    group1 = X_train.loc[y_train == 1, col].dropna()
    stat, p_value = mannwhitneyu(group0, group1, alternative="two-sided")
    mw_results.append({"variable": col, "median_no_default": group0.median(),
                        "median_default": group1.median(), "u_statistic": stat, "p_value": p_value})

mw_df = pd.DataFrame(mw_results)
mw_df["significant_bonferroni"] = mw_df["p_value"] < (0.05 / len(numeric_vars))
display(mw_df)
```

```python
def engineer_date_features(df):
    df = df.copy()
    issue_dt = pd.to_datetime(df["issue_d"], format="%b-%Y", errors="coerce")
    earliest_dt = pd.to_datetime(df["earliest_cr_line"], format="%b-%Y", errors="coerce")

    df["credit_history_months"] = (
        (issue_dt.dt.year - earliest_dt.dt.year) * 12
        + (issue_dt.dt.month - earliest_dt.dt.month)
    )
    issue_month = issue_dt.dt.month
    df["issue_month_sin"] = np.sin(2 * np.pi * issue_month / 12)
    df["issue_month_cos"] = np.cos(2 * np.pi * issue_month / 12)
    return df

X_train_fe = engineer_date_features(X_train)
X_test_fe = engineer_date_features(X_test)

print("X_train_fe shape:", X_train_fe.shape)
print("Missing in credit_history_months:", X_train_fe["credit_history_months"].isna().sum())
```

```python
flag_cols_final = [c for c in X_train_fe.columns if c.endswith("_was_missing")]
flag_corr = X_train_fe[flag_cols_final].corr().abs()

parent = {c: c for c in flag_cols_final}

def find(x):
    while parent[x] != x:
        parent[x] = parent[parent[x]]
        x = parent[x]
    return x

def union(x, y):
    rx, ry = find(x), find(y)
    if rx != ry:
        parent[ry] = rx

for i in range(len(flag_cols_final)):
    for j in range(i + 1, len(flag_cols_final)):
        c1, c2 = flag_cols_final[i], flag_cols_final[j]
        if flag_corr.loc[c1, c2] >= 0.95:
            union(c1, c2)

clusters = {}
for c in flag_cols_final:
    clusters.setdefault(find(c), []).append(c)

flags_to_keep = [min(members, key=len) for members in clusters.values()]
flags_to_drop = [c for c in flag_cols_final if c not in flags_to_keep]

X_train_final = X_train_fe.drop(columns=flags_to_drop)
X_test_final = X_test_fe.drop(columns=flags_to_drop)
flag_cols_final_reduced = [c for c in flag_cols_final if c not in flags_to_drop]

print("Clusters:", len(clusters), "| Flags kept:", len(flags_to_keep), "| Dropped:", len(flags_to_drop))
print("X_train_final shape:", X_train_final.shape)

del X_train_fe, X_test_fe
gc.collect()
```

```python
class_counts = y_train.value_counts().sort_index()
scale_pos_weight = class_counts[0] / class_counts[1]
print("Imbalance ratio (non-default : default): {:.2f} : 1".format(scale_pos_weight))
print("scale_pos_weight for XGBoost:", round(scale_pos_weight, 4))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score, average_precision_score, classification_report

date_cols = ["issue_d", "earliest_cr_line"]
sentinel_impute_cols = [c for c in high_missing_cols if c in X_train_final.columns]
median_impute_cols = [
    c for c in X_train_final.select_dtypes(include=np.number).columns
    if c not in sentinel_impute_cols and not c.endswith("_was_missing")
]
ordinal_cols = ["emp_length", "grade", "sub_grade"]
onehot_cols = [
    c for c in X_train_final.select_dtypes(include=["object", "category"]).columns
    if c not in ordinal_cols + date_cols
]

emp_length_order = ["< 1 year", "1 year", "2 years", "3 years", "4 years",
                     "5 years", "6 years", "7 years", "8 years", "9 years", "10+ years"]
grade_order = ["A", "B", "C", "D", "E", "F", "G"]
sub_grade_order = [f"{g}{n}" for g in grade_order for n in range(1, 6)]

sentinel_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="constant", fill_value=999)),
    ("scaler", StandardScaler())
])
median_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
ordinal_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OrdinalEncoder(categories=[emp_length_order, grade_order, sub_grade_order],
                                handle_unknown="use_encoded_value", unknown_value=-1))
])
onehot_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

preprocessor = ColumnTransformer([
    ("sentinel_num", sentinel_pipeline, sentinel_impute_cols),
    ("median_num", median_pipeline, median_impute_cols),
    ("flags", "passthrough", flag_cols_final_reduced),
    ("ordinal_cat", ordinal_pipeline, ordinal_cols),
    ("onehot_cat", onehot_pipeline, onehot_cols),
], remainder="drop")

logreg_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(class_weight="balanced", max_iter=1000, random_state=42))
])

xgb_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", XGBClassifier(
        scale_pos_weight=scale_pos_weight, n_estimators=300, max_depth=5,
        learning_rate=0.05, subsample=0.8, colsample_bytree=0.8,
        eval_metric="auc", random_state=42, n_jobs=-1
    ))
])

print("Fitting Logistic Regression...")
logreg_pipeline.fit(X_train_final, y_train)
print("Fitting XGBoost...")
xgb_pipeline.fit(X_train_final, y_train)

for name, pipe in [("Logistic Regression", logreg_pipeline), ("XGBoost", xgb_pipeline)]:
    train_proba = pipe.predict_proba(X_train_final)[:, 1]
    train_pred = pipe.predict(X_train_final)
    print(f"\n=== {name} — train sanity check ===")
    print("ROC-AUC:", round(roc_auc_score(y_train, train_proba), 4))
    print("PR-AUC:", round(average_precision_score(y_train, train_proba), 4))
    print(classification_report(y_train, train_pred, digits=3))
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate

skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scoring = {"roc_auc": "roc_auc", "pr_auc": "average_precision"}

cv_results_summary = {}
for name, pipe in [("Logistic Regression", logreg_pipeline), ("XGBoost", xgb_pipeline)]:
    print(f"\nRunning 5-fold CV for {name}...")
    cv_results = cross_validate(pipe, X_train_final, y_train, cv=skf, scoring=scoring, n_jobs=1)
    cv_results_summary[name] = {
        "roc_auc_mean": cv_results["test_roc_auc"].mean(),
        "roc_auc_std": cv_results["test_roc_auc"].std(),
        "pr_auc_mean": cv_results["test_pr_auc"].mean(),
        "pr_auc_std": cv_results["test_pr_auc"].std(),
    }
    print("ROC-AUC: {:.4f} +/- {:.4f}".format(cv_results["test_roc_auc"].mean(), cv_results["test_roc_auc"].std()))
    print("PR-AUC: {:.4f} +/- {:.4f}".format(cv_results["test_pr_auc"].mean(), cv_results["test_pr_auc"].std()))

cv_summary_df = pd.DataFrame([{"Model": name, **vals} for name, vals in cv_results_summary.items()])
display(cv_summary_df)
```

```python
from sklearn.model_selection import RandomizedSearchCV

cv_light = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)

logreg_search = RandomizedSearchCV(
    logreg_pipeline, param_distributions={"classifier__C": [0.01, 0.1, 1, 10]},
    n_iter=4, scoring="roc_auc", cv=cv_light, random_state=42, n_jobs=1, verbose=1
)
print("Tuning Logistic Regression...")
logreg_search.fit(X_train_final, y_train)
print("Best C:", logreg_search.best_params_, "| CV ROC-AUC:", round(logreg_search.best_score_, 4))

xgb_search = RandomizedSearchCV(
    xgb_pipeline,
    param_distributions={
        "classifier__max_depth": [4, 5, 6],
        "classifier__learning_rate": [0.03, 0.05, 0.1],
        "classifier__n_estimators": [200, 300, 400],
        "classifier__min_child_weight": [1, 5, 10]
    },
    n_iter=8, scoring="roc_auc", cv=cv_light, random_state=42, n_jobs=1, verbose=1
)
print("\nTuning XGBoost...")
xgb_search.fit(X_train_final, y_train)
print("Best params:", xgb_search.best_params_, "| CV ROC-AUC:", round(xgb_search.best_score_, 4))

best_logreg = logreg_search.best_estimator_
best_xgb = xgb_search.best_estimator_

save_checkpoint("tuned_models", best_logreg=best_logreg, best_xgb=best_xgb,
                preprocessor=preprocessor, X_train_final=X_train_final, X_test_final=X_test_final,
                y_train=y_train, y_test=y_test, scale_pos_weight=scale_pos_weight)
```

```python
comparison_df = pd.DataFrame([
    {"Model": "Logistic Regression (tuned)", "Best CV ROC-AUC": round(logreg_search.best_score_, 4)},
    {"Model": "XGBoost (tuned)", "Best CV ROC-AUC": round(xgb_search.best_score_, 4)},
])
display(comparison_df)

print("""
Decision: XGBoost (tuned) carries forward as the primary model for
calibration, threshold analysis, and locked test evaluation.
Logistic Regression is retained as an interpretable benchmark for
the odds-ratio analysis, since its coefficients are directly interpretable.
""")
```

```python
preprocessor.fit(X_train_final)
X_train_prepared = preprocessor.transform(X_train_final)
X_test_prepared = preprocessor.transform(X_test_final)
feature_names = preprocessor.get_feature_names_out()

print("X_train_prepared shape:", X_train_prepared.shape)
print("X_test_prepared shape:", X_test_prepared.shape)
```

```python
from sklearn.model_selection import train_test_split as tts
from sklearn.calibration import CalibratedClassifierCV, calibration_curve
from sklearn.metrics import brier_score_loss

X_fit, X_calib, y_fit, y_calib = tts(
    X_train_prepared, y_train, test_size=0.20, stratify=y_train, random_state=42
)
print("X_fit:", X_fit.shape, "| X_calib:", X_calib.shape)

best_xgb_params = best_xgb.named_steps["classifier"].get_params()
tuned_kwargs = {k: best_xgb_params[k] for k in ["max_depth", "learning_rate", "n_estimators", "min_child_weight"]}

xgb_for_calib = XGBClassifier(
    scale_pos_weight=scale_pos_weight, eval_metric="auc", random_state=42,
    n_jobs=1, tree_method="hist", **tuned_kwargs
)
xgb_for_calib.fit(X_fit, y_fit)

calibrated_xgb = CalibratedClassifierCV(xgb_for_calib, method="isotonic", cv="prefit")
calibrated_xgb.fit(X_calib, y_calib)

raw_proba = xgb_for_calib.predict_proba(X_calib)[:, 1]
calib_proba = calibrated_xgb.predict_proba(X_calib)[:, 1]

frac_pos_raw, mean_pred_raw = calibration_curve(y_calib, raw_proba, n_bins=10, strategy="quantile")
frac_pos_calib, mean_pred_calib = calibration_curve(y_calib, calib_proba, n_bins=10, strategy="quantile")

plt.figure(figsize=(7, 7))
plt.plot([0, 1], [0, 1], "k--", label="Perfectly calibrated")
plt.plot(mean_pred_raw, frac_pos_raw, "o-", label="Raw XGBoost")
plt.plot(mean_pred_calib, frac_pos_calib, "o-", label="Calibrated (isotonic)")
plt.xlabel("Mean predicted probability")
plt.ylabel("Fraction of actual positives")
plt.title("Calibration curve — raw vs calibrated XGBoost")
plt.legend()
plt.show()

print("Brier score (raw):", round(brier_score_loss(y_calib, raw_proba), 5))
print("Brier score (calibrated):", round(brier_score_loss(y_calib, calib_proba), 5))
```

```python
from sklearn.metrics import precision_recall_curve, confusion_matrix

precisions, recalls, thresholds = precision_recall_curve(y_calib, calib_proba)
f1_scores = 2 * (precisions[:-1] * recalls[:-1]) / (precisions[:-1] + recalls[:-1] + 1e-9)
best_idx = np.argmax(f1_scores)
best_threshold_f1 = thresholds[best_idx]

print(f"Best threshold by F1: {best_threshold_f1:.4f}")
print(f"Precision: {precisions[best_idx]:.4f} | Recall: {recalls[best_idx]:.4f} | F1: {f1_scores[best_idx]:.4f}")

plt.figure(figsize=(9, 6))
plt.plot(thresholds, precisions[:-1], label="Precision")
plt.plot(thresholds, recalls[:-1], label="Recall")
plt.plot(thresholds, f1_scores, label="F1")
plt.axvline(best_threshold_f1, color="red", linestyle="--", label=f"Best F1 threshold={best_threshold_f1:.3f}")
plt.axvline(0.5, color="gray", linestyle=":", label="Default 0.5")
plt.xlabel("Decision threshold")
plt.ylabel("Score")
plt.title("Threshold analysis (on calibration set)")
plt.legend()
plt.show()

for thresh, label in [(0.5, "Default 0.5"), (best_threshold_f1, "Best F1 threshold")]:
    preds = (calib_proba >= thresh).astype(int)
    cm = confusion_matrix(y_calib, preds)
    print(f"\n=== {label} (threshold={thresh:.3f}) ===")
    print("Confusion matrix [[TN, FP], [FN, TP]]:")
    print(cm)
```

```python
from sklearn.metrics import roc_auc_score, average_precision_score, confusion_matrix, classification_report

test_proba = calibrated_xgb.predict_proba(X_test_prepared)[:, 1]
LOCKED_THRESHOLD = round(best_threshold_f1, 3)  # decided on calibration data only
test_preds = (test_proba >= LOCKED_THRESHOLD).astype(int)

print("=" * 60)
print("LOCKED TEST SET EVALUATION — FINAL RESULT")
print("=" * 60)

roc_auc = roc_auc_score(y_test, test_proba)
pr_auc = average_precision_score(y_test, test_proba)
brier = brier_score_loss(y_test, test_proba)

print(f"\nROC-AUC: {roc_auc:.4f}")
print(f"PR-AUC: {pr_auc:.4f}")
print(f"Brier score: {brier:.5f}")

cm = confusion_matrix(y_test, test_preds)
print(f"\nConfusion matrix at threshold={LOCKED_THRESHOLD} [[TN, FP], [FN, TP]]:")
print(cm)
print(f"\nClassification report (threshold={LOCKED_THRESHOLD}):")
print(classification_report(y_test, test_preds, digits=3))

cv_xgb_roc_auc = cv_results_summary["XGBoost"]["roc_auc_mean"]
print("\n=== Comparison: CV (train) vs Locked Test ===")
print(f"CV ROC-AUC (train): {cv_xgb_roc_auc:.4f}")
print(f"Test ROC-AUC: {roc_auc:.4f}")
print("(should be close — a large gap would suggest the CV estimate was optimistic)")

print("\nTest set evaluation COMPLETE and LOCKED. Do not re-run with a different threshold.")
```
